```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pth = "/Users/yilewang/workspaces/data4project/"
colors = ["#66CDAA","#4682B4","#AB63FA","#FFA15A"]
# load data
df = pd.read_excel(pth + "/prediction_project/prediction_data.xlsx", sheet_name="main_spike_burst", skiprows=1)
# plot all the columns
for i in range(5, len(df.columns)):
    figure = plt.figure()
    # hide the top and right spines
    ax = plt.gca()
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    sns.violinplot(x="group", y=df.columns[i], data=df, inner=None, palette=colors)
    sns.pointplot(x="group", y=df.columns[i], data=df, color='r', ci=None)
    sns.stripplot(x="group", y=df.columns[i], data=df, color='black', alpha=0.5,zorder=1)
    plt.show()
```
